Output usable mapping for the DeBERTa

```python
import pprint

cols = ['Original_Label', 'Primary_Emotion_1', 'Primary_Emotion_2']

mappings = df[cols].apply(lambda col: col.str.lower()).drop_duplicates()

result = {}
for row in mappings.itertuples(index=False):
  key = str(row.Original_Label)
  result.setdefault(key, []).append(str(row.Primary_Emotion_1))
  result[key].append(None if str(row.Primary_Emotion_2) == "nan" else str(row.Primary_Emotion_2))
print(f"data = {pprint.pformat(result, sort_dicts=False)}")

```
