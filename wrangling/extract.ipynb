```python
import pandas as pd

messy = pd.read_csv('concatenated.csv')
messy
```

```python
extracted = messy[['Source_Dataset', 'Text', 'Primary_Emotion_1']]
extracted = extracted[extracted['Source_Dataset'] == 'EmoLex']

label_cols = ['Joy', 'Trust', 'Fear', 'Surprise', 'Sadness', 'Disgust', 'Anger', 'Anticipation', 'Neutral']

print(sorted(extracted["Primary_Emotion_1"].unique()))

wide = (
    pd.get_dummies(extracted["Primary_Emotion_1"])
      .reindex(columns=label_cols, fill_value=0)
      .astype(int)
      .join(extracted[["Source_Dataset", "Text"]])
      .groupby(["Source_Dataset", "Text"], as_index=False)
      .max()
)

wide = wide.rename(columns={"Text": "text", "Source_Dataset": "source"})

wide.to_csv('emolex_wide.csv')
```
